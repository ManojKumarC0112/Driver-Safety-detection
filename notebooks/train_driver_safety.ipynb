## 4. Model Architecture & Latency Benchmark

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = DriverSafetyNet(
    in_channels=12,
    cnn_filters=32,
    kernel_size=3,
    lstm_hidden_size=64,
    lstm_num_layers=2,
    lstm_bidirectional=True,
    num_classes=4
).to(device)

# Calculate total trainable parameters
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Device: {device}")
print(f"Total Trainable Parameters: {num_params:,}")

# Test latency benchmark on CPU/GPU
dummy_input = torch.randn(1, 30, 12).to(device)
model.eval()
with torch.no_grad():
    for _ in range(10): _ = model(dummy_input) # Warmup
    t0 = time.time()
    for _ in range(100): _ = model(dummy_input)
    avg_latency_ms = ((time.time() - t0) / 100.0) * 1000.0

print(f"Model Inference Latency: {avg_latency_ms:.3f} ms / frame")
```
